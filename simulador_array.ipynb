# 🗄️ Simulador de Arrays em Python
Este projeto utiliza a analogia de **gavetas** para demonstrar como funcionam as operações em **Arrays/Listas**.

```python
# 1. Criando o Array (Armário de tamanho fixo com 5 posições)
TAMANHO = 5
armario = [None] * TAMANHO

def exibir_armario():
    print("\n--- ESTADO ATUAL DO ARRAY ---")
    for indice in range(TAMANHO):
        conteudo = armario[indice]
        status = f"[ {conteudo} ]" if conteudo is not None else "[ VAZIO ]"
        print(f"Índice {indice}: {status}")
    print("-----------------------------\n")

def guardar_item(indice, item):
    if 0 <= indice < TAMANHO:
        armario[indice] = item
        print(f"✅ Item '{item}' guardado no índice {indice}.")
    else:
        print(f"❌ Erro: ÍNDICE {indice} FORA DOS LIMITES DO ARRAY!")

def esvaziar_gaveta(indice):
    if 0 <= indice < TAMANHO:
        removido = armario[indice]
        armario[indice] = None
        print(f"🗑️ Item '{removido}' removido do índice {indice}.")
    else:
        print(f"❌ Erro: ÍNDICE {indice} INVALIDO!")

def buscar_item(item):
    print(f"🔍 Buscando '{item}' no array...")
    # Busca linear no array
    for i in range(TAMANHO):
        if armario[i] is not None and armario[i].lower() == item.lower():
            print(f"🎯 Encontrado! O item '{armario[i]}' está no índice {i}.")
            return i
    print(f"❌ O item '{item}' não existe no array.")
    return -1
```

```python
# 2. Testando as Operações no Array
exibir_armario()

# Guardando elementos por índice
guardar_item(0, "Camisa")
guardar_item(2, "Livro")
exibir_armario()

# Buscando um item pelo valor
buscar_item("Livro")

# Removendo elemento
esvaziar_gaveta(0)
exibir_armario()
```
